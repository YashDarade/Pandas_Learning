```python
import numpy as np
import pandas as pd
```

```python
## Loading the data
df = pd.read_csv(r"anime.csv")
```

```python
df.head()
```

### Make a new column for episode count

```python

df.loc[1]['Title']

## as we can see there are episode counts in the title 
```

```python
def extract_episode(txt):
    check = False
    data = ""
    for i in txt:
        if i == ')':
            break

        if check == True:
            data += i

        if i == '(':
            check = True
            
    return data
        
```

```python
df["Episodes"] = df['Title'].apply(extract_episode)
```

```python
df['Episodes'] = df['Episodes'].str.replace(" eps","")
```

```python
df['Episodes'] = df['Episodes'].astype(int)
df.head()
```

### Make a new column for time stamp

```python
df.loc[0]['Title']
```

```python
def extraction_time(txt):
    check = False
    data = ""
    for i in range(len(txt)):
        if txt[i]==')':
            for j in range(i+1,i+20):
                data += txt[j]

    return data

```

```python
df['Total_time'] = df['Title'].apply(extraction_time)

df.head()
```

```python
## Adding total months 
def calculate_total_months(duration):
    """
    Calculate the total number of months from a string like:
    'Apr 2009 - Jul 2010'
    """
    start_str, end_str = duration.split(" - ")

    start = pd.to_datetime(start_str, format="%b %Y")
    end = pd.to_datetime(end_str, format="%b %Y")

    total_months = (
        (end.year - start.year) * 12 +
        (end.month - start.month) + 1
    )

    return total_months
```

```python
df['Months'] = df['Total_time'].apply(calculate_total_months)
df.head()
```
